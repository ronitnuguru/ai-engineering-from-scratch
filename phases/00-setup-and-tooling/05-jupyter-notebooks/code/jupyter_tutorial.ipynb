## 2. Magic Commands

### Timing with %timeit (runs multiple times)

```python
# Compare list comprehension vs NumPy
size = 100000

print("List comprehension:")
%timeit [np.random.random() for _ in range(size)]
```

```python
print("NumPy (vectorized):")
%timeit np.random.random(size)
```

**NumPy is ~100x faster!** This is why we use it for arrays.

### Timing with %%time (single run for larger operations)

```python
%%time
# Simulate a training loop
total = 0
for i in range(1000000):
    total += i
print(f"Result: {total}")
```

## 3. Inline Visualization

Enable matplotlib inline rendering:

```python
%matplotlib inline

# Create sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Plot
plt.figure(figsize=(10, 4))
plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, label='cos(x)')
plt.title('Inline Plotting Demo')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

## 4. Rich Display with Pandas

```python
import pandas as pd

# Create sample dataset
data = {
    'Model': ['Linear Regression', 'Random Forest', 'Neural Network', 'XGBoost'],
    'Accuracy': [0.72, 0.89, 0.94, 0.91],
    'Training Time (s)': [0.1, 2.3, 45.6, 8.4],
    'Parameters': ['2', '100 trees', '1.2M', '1000']
}

df = pd.DataFrame(data)
df
```

```python
# Visualize the comparison
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Accuracy comparison
axes[0].bar(df['Model'], df['Accuracy'])
axes[0].set_title('Model Accuracy')
axes[0].set_ylabel('Accuracy')
axes[0].tick_params(axis='x', rotation=45)

# Training time comparison (log scale)
axes[1].bar(df['Model'], df['Training Time (s)'])
axes[1].set_title('Training Time')
axes[1].set_ylabel('Time (seconds, log scale)')
axes[1].set_yscale('log')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

## 9. Common Traps to Avoid

### Trap 1: Out-of-order execution
Run cells top-to-bottom before sharing. Use: **Kernel → Restart & Run All**

### Trap 2: Hidden state
Delete a cell but the variable remains in memory. Restart kernel to clear.

### Trap 3: Memory leaks
Large datasets accumulate. Use `del variable` and `import gc; gc.collect()`
